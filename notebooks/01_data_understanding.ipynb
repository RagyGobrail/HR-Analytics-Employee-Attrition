```python
categorical_columns = df.select_dtypes(
    include="object"
).columns

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

```python
checks = {
    "Invalid age": (df["Age"] <= 0).sum(),
    "Invalid monthly income": (df["MonthlyIncome"] <= 0).sum(),
    "Negative years at company": (df["YearsAtCompany"] < 0).sum(),
    "Negative total working years": (df["TotalWorkingYears"] < 0).sum(),
    "Invalid attrition values": (~df["Attrition"].isin(["Yes", "No"])).sum()
}

for check, count in checks.items():
    print(f"{check}: {count}")
```

```python
df_clean = df.copy()
```

```python
columns_to_remove = [
    "EmployeeCount",
    "Over18",
    "StandardHours"
]

df_clean = df_clean.drop(
    columns=columns_to_remove
)
```

```python
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
```
